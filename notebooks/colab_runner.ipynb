```python
import json
import pandas as pd
from pathlib import Path

runs = [
    "standard",
    "tsa_sigma_0_1",
    "tsa_sigma_1_0",
    "tsa_sigma_3_0",
]

rows = []

for run in runs:
    row = {"run": run}

    train_path = Path("outputs") / run / "metrics.json"
    snap_path = Path("outputs") / run / "eval_snapshot_metrics.json"
    probe_path = Path("outputs") / run / "eval_probe_metrics.json"

    if train_path.exists():
        with open(train_path, "r") as f:
            d = json.load(f)
        row["final_loss"] = d.get("final_loss")
        row["final_loss_data"] = d.get("final_loss_data")
        row["final_loss_res"] = d.get("final_loss_res")
        row["final_loss_test"] = d.get("final_loss_test")
        row["final_loss_reg"] = d.get("final_loss_reg")

    if snap_path.exists():
        with open(snap_path, "r") as f:
            d = json.load(f)
        row["snapshot_rel_l2_u"] = d.get("relative_l2_error_u")
        row["snapshot_rel_l2_v"] = d.get("relative_l2_error_v")

    if probe_path.exists():
        with open(probe_path, "r") as f:
            d = json.load(f)
        row["probe_rel_l2_u"] = d.get("relative_l2_error_u_probe")
        row["probe_rel_l2_v"] = d.get("relative_l2_error_v_probe")

    rows.append(row)

df_compare = pd.DataFrame(rows)
df_compare
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

runs = [
    "standard",
    "tsa_sigma_0_1",
    "tsa_sigma_1_0",
    "tsa_sigma_3_0",
]

plt.figure(figsize=(10, 6))

for run in runs:
    history_path = Path("outputs") / run / "history.csv"
    if history_path.exists():
        df = pd.read_csv(history_path)
        plt.plot(df["epoch"], df["loss"], label=run)

plt.yscale("log")
plt.xlabel("Epoch")
plt.ylabel("Total Loss")
plt.title("Training Loss Comparison")
plt.legend()
plt.grid(True, linestyle="--", alpha=0.5)
plt.show()
```

```python
plt.figure(figsize=(10, 6))

for run in runs:
    history_path = Path("outputs") / run / "history.csv"
    if history_path.exists():
        df = pd.read_csv(history_path)
        plt.plot(df["epoch"], df["loss_test"], label=run)

plt.yscale("log")
plt.xlabel("Epoch")
plt.ylabel("Held-out Data Loss")
plt.title("Test Loss Comparison")
plt.legend()
plt.grid(True, linestyle="--", alpha=0.5)
plt.show()
```

```python
plt.figure(figsize=(10, 5))
plt.bar(df_compare["run"], df_compare["snapshot_rel_l2_u"])
plt.ylabel("Relative L2 Error")
plt.title("Snapshot Relative L2 Error in u")
plt.xticks(rotation=20)
plt.show()

plt.figure(figsize=(10, 5))
plt.bar(df_compare["run"], df_compare["probe_rel_l2_u"])
plt.ylabel("Relative L2 Error")
plt.title("Probe Relative L2 Error in u(t)")
plt.xticks(rotation=20)
plt.show()
```
